```python
from google.colab import userdata
API_KEY = userdata.get("Riya")
print("API key loaded successfully!")
```

```python
from google import genai
client = genai.Client(api_key=API_KEY)
print("Gemini connected successfully!")
```

```python
response = client.models.generate_content(
    model="gemini-3.5-flash",
    contents="Hello! Introduce yourself as an emotionally intelligent AI assistant."
)

print(response.text)
```

```python
def chatbot(message):
    prompt = f"""
You are an emotionally intelligent AI chatbot.

Understand the emotion in the user's message and reply appropriately.
If the user is happy, be cheerful.
If the user is sad, be supportive.
If the user is angry, be calm.
If the user is stressed, be encouraging.
If the user asks a normal question, answer it normally.

User message: {message}
"""

    response = client.models.generate_content(
        model="gemini-3.1-flash-lite",
        contents=prompt
    )

    return response.text


print(chatbot("I am very stressed about my exam tomorrow."))
```

```python
import gradio as gr

def chat(message, history):
    prompt = f"""
You are an emotionally intelligent AI chatbot.

First identify the main emotion in the user's message.
Choose one: Happy, Sad, Angry, Stressed, Excited, Anxious, Tired, or Neutral.

Then answer the user appropriately and adapt your tone to their emotion.

Give your answer in this exact format:

Emotion: [one emotion]
Response: [your response]

User message: {message}
"""

    response = client.models.generate_content(
        model="gemini-3.1-flash-lite",
        contents=prompt
    )

    return response.text


demo = gr.ChatInterface(
    fn=chat,
    chatbot=gr.Chatbot(height=300),
    title="💙 Emotionally Intelligent AI Chatbot",
    description="Powered by Gemini Flash",
    fill_height=False
)

demo.launch()
```
